````python
def get_test_values(n=10):
   
    import numpy as np

    # Create an instance of a random number generator
    rng = np.random.default_rng(seed=1234)

    # Generate n random values uniformly distributed in the range [0, 1]
    values = rng.random(n)

    return values
````
